```python
## Experiment A — Classification (predict escalated)
## 1. Loading and inspecting the data

import warnings
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, StratifiedKFold, KFold, cross_validate, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression, LinearRegression, Ridge
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.dummy import DummyClassifier, DummyRegressor
from sklearn.metrics import (classification_report, confusion_matrix, ConfusionMatrixDisplay,
                             mean_absolute_error, mean_squared_error, r2_score)

warnings.filterwarnings("ignore")
RANDOM_STATE = 42          # one seed everywhere -> reproducible results
pd.set_option("display.width", 160)
gression data:", reg_df.shape)
print("\nMissing values per column (classification):")
print(clf_df.isna().sum())
print("\nTarget balance (escalated):")
print(clf_df["escalated"].value_counts(norm
clf_df = pd.read_csv("data/classification_escalation.csv")
reg_df = pd.read_csv("data/resolution_time_regression.csv")
display(clf_df.head(), reg_df.head())

print("Classification data:", clf_df.shape, "| Realize=True).round(3))
print("\nticket_id unique per row? ->", clf_df["ticket_id"].is_unique, "(it is just an identifier, so we drop it)")

## 2. Stratified train/test split

# Drop the identifier and separate the target
X = clf_df.drop(columns=["ticket_id", "escalated"])
y = clf_df["escalated"]

# Stratified hold-out split: the test set is locked away until the very end
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=RANDOM_STATE)

print("Train:", X_train.shape, " Test:", X_test.shape)
print("Share of escalated=1  -> full: %.3f | train: %.3f | test: %.3f"
      % (y.mean(), y_train.mean(), y_test.mean()))

## 3. Preprocessing (inside a pipeline)

num_cols = X.select_dtypes(include="number").columns.tolist()
cat_cols = X.select_dtypes(exclude="number").columns.tolist()
print("Numeric:", num_cols)
print("Categorical:", cat_cols)

def make_preprocessor():
    """A *fresh* preprocessor each time, so no fitted state is ever shared between models."""
    numeric = Pipeline([("impute", SimpleImputer(strategy="median")),
                        ("scale", StandardScaler())])
    categorical = Pipeline([("impute", SimpleImputer(strategy="most_frequent")),
                            ("onehot", OneHotEncoder(handle_unknown="ignore"))])
    return ColumnTransformer([("num", numeric, num_cols),
                              ("cat", categorical, cat_cols)])

make_preprocessor()

## 4. Comparing three models with identical 5-fold stratified CV

cv_clf = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)  # identical folds for every model

candidates = {
    "Majority-class baseline": DummyClassifier(strategy="most_frequent"),
    "Logistic Regression":     LogisticRegression(max_iter=1000, random_state=RANDOM_STATE),
    "k-NN":                    KNeighborsClassifier(),
    "Decision Tree":           DecisionTreeClassifier(random_state=RANDOM_STATE),
}
scoring = {"f1": "f1", "precision": "precision", "recall": "recall", "accuracy": "accuracy"}

rows, fold_f1 = [], {}
for name, model in candidates.items():
    pipe = Pipeline([("prep", make_preprocessor()), ("model", model)])
    res = cross_validate(pipe, X_train, y_train, cv=cv_clf, scoring=scoring)
    fold_f1[name] = res["test_f1"]
    rows.append({"Model": name,
                 "F1 mean": res["test_f1"].mean(), "F1 std": res["test_f1"].std(),
                 "Precision": res["test_precision"].mean(), "Recall": res["test_recall"].mean(),
                 "Accuracy": res["test_accuracy"].mean()})

cv_table_A = pd.DataFrame(rows).set_index("Model").round(3)
display(cv_table_A)
print("Per-fold F1 scores:")
for k, v in fold_f1.items():
    print(f"  {k:24s}", np.round(v, 3))

## 5. Hyper-parameter tuning (GridSearchCV)

param_grids = {
    "Logistic Regression": {"model__C": [0.01, 0.1, 1, 10, 100],
                            "model__class_weight": [None, "balanced"]},
    "k-NN":                {"model__n_neighbors": [3, 5, 7, 11, 15, 21, 31],
                            "model__weights": ["uniform", "distance"]},
    "Decision Tree":       {"model__max_depth": [2, 3, 4, 5, 6, 8, None],
                            "model__min_samples_leaf": [1, 5, 10, 20],
                            "model__class_weight": [None, "balanced"]},
}
base_models = {k: candidates[k] for k in param_grids}

searches, tune_rows = {}, []
for name, grid in param_grids.items():
    pipe = Pipeline([("prep", make_preprocessor()), ("model", base_models[name])])
    gs = GridSearchCV(pipe, grid, scoring="f1", cv=cv_clf, n_jobs=-1, refit=True)
    gs.fit(X_train, y_train)                      # training data only
    searches[name] = gs
    i = gs.best_index_
    tune_rows.append({"Model": name,
                      "Best params": {k.replace("model__", ""): v for k, v in gs.best_params_.items()},
                      "CV F1 mean": gs.cv_results_["mean_test_score"][i],
                      "CV F1 std":  gs.cv_results_["std_test_score"][i]})

tuned_table_A = pd.DataFrame(tune_rows).set_index("Model")
tuned_table_A[["CV F1 mean", "CV F1 std"]] = tuned_table_A[["CV F1 mean", "CV F1 std"]].round(3)
pd.set_option("display.max_colwidth", 100)
display(tuned_table_A)

# Per-fold F1 for the tuned models (same folds for each)
tuned_fold_f1 = {}
for name, gs in searches.items():
    res = cross_validate(gs.best_estimator_, X_train, y_train, cv=cv_clf, scoring="f1")
    tuned_fold_f1[name] = res["test_score"]

fig, ax = plt.subplots(figsize=(7, 3.8))
ax.boxplot(list(tuned_fold_f1.values()), showmeans=True)
ax.set_xticks(range(1, len(tuned_fold_f1) + 1)); ax.set_xticklabels(list(tuned_fold_f1.keys()))
for i, v in enumerate(tuned_fold_f1.values(), start=1):
    ax.scatter(np.full(len(v), i), v, alpha=.6, zorder=3)
ax.set_ylabel("F1 per fold"); ax.set_title("Tuned models: F1 across the 5 CV folds")
plt.tight_layout(); plt.show()

## 6. Final evaluation on the test set

# Pick the winner by CV F1 (never by test score)
best_name_A = tuned_table_A["CV F1 mean"].astype(float).idxmax()
final_clf = searches[best_name_A].best_estimator_
print("Selected model:", best_name_A, searches[best_name_A].best_params_)

# ---- The one and only look at the test set ----
y_pred = final_clf.predict(X_test)
print("\n", classification_report(y_test, y_pred, target_names=["not escalated", "escalated"], digits=3))

tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()
print(f"TN={tn}  FP={fp}  FN={fn}  TP={tp}")

fig, ax = plt.subplots(figsize=(4.2, 3.8))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred, display_labels=["no", "yes"], cmap="Blues", ax=ax, colorbar=False)
ax.set_title("Test-set confusion matrix"); plt.tight_layout(); plt.show()

## Experiment B — Regression (predict resolution_time_hours)
## 7. Inspecting the target and outliers

print(reg_df.isna().sum())
y_all = reg_df["resolution_time_hours"]
print("\n", y_all.describe(percentiles=[.5, .9, .95, .99]).round(2))

q1, q3 = y_all.quantile([.25, .75]); upper = q3 + 1.5 * (q3 - q1)
print(f"\nIQR rule upper fence = {upper:.2f} h -> {(y_all > upper).sum()} tickets above it")
print(f"Tickets sitting exactly on the 0.5 h floor: {(y_all == 0.5).sum()} ({(y_all == 0.5).mean():.1%})")

fig, axes = plt.subplots(1, 2, figsize=(9, 3.2))
axes[0].hist(y_all, bins=40); axes[0].set_title("resolution_time_hours"); axes[0].set_xlabel("hours")
axes[1].boxplot(y_all, vert=False); axes[1].set_title("Boxplot (outliers = dots)")
plt.tight_layout(); plt.show()

Xr = reg_df.drop(columns=["ticket_id", "resolution_time_hours"])
yr = reg_df["resolution_time_hours"]

# Plain random split (stratification is for class labels; the target here is continuous)
Xr_train, Xr_test, yr_train, yr_test = train_test_split(Xr, yr, test_size=0.20, random_state=RANDOM_STATE)

num_cols = Xr.select_dtypes(include="number").columns.tolist()      # make_preprocessor() reads these
cat_cols = Xr.select_dtypes(exclude="number").columns.tolist()
print("Train:", Xr_train.shape, " Test:", Xr_test.shape)

## 8. Comparing three regressors with 5-fold CV (MAE)

cv_reg = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
reg_models = {
    "Median baseline":   DummyRegressor(strategy="median"),
    "Linear Regression": LinearRegression(),
    "Ridge":             Ridge(random_state=RANDOM_STATE),
    "Decision Tree":     DecisionTreeRegressor(random_state=RANDOM_STATE),
}
scoring_r = {"mae": "neg_mean_absolute_error", "rmse": "neg_root_mean_squared_error"}

rows = []
for name, model in reg_models.items():
    pipe = Pipeline([("prep", make_preprocessor()), ("model", model)])
    res = cross_validate(pipe, Xr_train, yr_train, cv=cv_reg, scoring=scoring_r)
    mae, rmse = -res["test_mae"], -res["test_rmse"]
    rows.append({"Model": name, "MAE mean": mae.mean(), "MAE std": mae.std(),
                 "RMSE mean": rmse.mean(), "RMSE std": rmse.std()})
cv_table_B = pd.DataFrame(rows).set_index("Model").round(3)
display(cv_table_B)

## 9. Tuning ridge α and tree depth

param_grids_r = {
    "Ridge":         {"model__alpha": [0.01, 0.1, 1, 10, 30, 100, 300, 1000]},
    "Decision Tree": {"model__max_depth": [2, 3, 4, 5, 6, 8, 10, None],
                      "model__min_samples_leaf": [1, 5, 10, 20, 40]},
}
searches_r, rows = {}, []
for name, grid in param_grids_r.items():
    pipe = Pipeline([("prep", make_preprocessor()), ("model", reg_models[name])])
    gs = GridSearchCV(pipe, grid, scoring="neg_mean_absolute_error", cv=cv_reg, n_jobs=-1, refit=True)
    gs.fit(Xr_train, yr_train)
    searches_r[name] = gsf
    # Re-score the tuned pipeline on the same folds to also get per-fold MAE and RMSE
    res = cross_validate(gs.best_estimator_, Xr_train, yr_train, cv=cv_reg, scoring=scoring_r)
    rows.append({"Model": name + " (tuned)",
                 "Best params": {k.replace("model__", ""): v for k, v in gs.best_params_.items()},
                 "MAE mean": -res["test_mae"].mean(), "MAE std": res["test_mae"].std(),
                 "RMSE mean": -res["test_rmse"].mean(), "RMSE std": res["test_rmse"].std()})
    searches_r[name].fold_mae_ = -res["test_mae"]

tuned_table_B = pd.DataFrame(rows).set_index("Model")
tuned_table_B[["MAE mean", "MAE std", "RMSE mean", "RMSE std"]] = tuned_table_B[["MAE mean", "MAE std", "RMSE mean", "RMSE std"]].astype(float).round(3)
display(tuned_table_B)
for k, gs in searches_r.items():
    print(f"{k:14s} per-fold MAE:", np.round(gs.fold_mae_, 3))

## 10. Final test evaluation (once)

best_name_B = tuned_table_B["MAE mean"].astype(float).idxmin().replace(" (tuned)", "")
final_reg = searches_r[best_name_B].best_estimator_
print("Selected model:", best_name_B, searches_r[best_name_B].best_params_)

# ---- The one and only look at the regression test set ----
pred = final_reg.predict(Xr_test)
mae  = mean_absolute_error(yr_test, pred)
rmse = np.sqrt(mean_squared_error(yr_test, pred))
print(f"\nTEST  MAE = {mae:.3f} h | RMSE = {rmse:.3f} h | R2 = {r2_score(yr_test, pred):.3f}")

resid = yr_test - pred
big = resid.abs().sort_values(ascending=False).head(5)
print("\nLargest absolute errors (hours):"); print(big.round(2).to_string())

fig, axes = plt.subplots(1, 2, figsize=(9.5, 3.8))
axes[0].scatter(yr_test, pred, alpha=.6); lim = [0, max(yr_test.max(), pred.max()) + 1]
axes[0].plot(lim, lim, "r--"); axes[0].set_xlabel("actual hours"); axes[0].set_ylabel("predicted hours"); axes[0].set_title("Predicted vs actual (test)")
axes[1].hist(resid, bins=30); axes[1].set_title("Test residuals (actual - predicted)"); axes[1].set_xlabel("hours")
plt.tight_layout(); plt.show()

```
